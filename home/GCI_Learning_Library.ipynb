````python
# ============================================================
# GCI WORLD 2026 — ENVIRONMENT CONFIGURATION
# ============================================================

import sys
import subprocess
from pathlib import Path

import pandas as pd
import ipywidgets as widgets
from IPython.display import display, Markdown, HTML, clear_output


# ------------------------------------------------------------
# Project configuration
# ------------------------------------------------------------

REPO_NAME = "GCI-World-2026-September"

REPO_URL = (
    "https://github.com/"
    "GabrielRamos1171998/"
    "GCI-World-2026-September.git"
)


# ------------------------------------------------------------
# Detect project root
# ------------------------------------------------------------

current = Path.cwd()

possible_roots = [
    current,
    *current.parents
]

ROOT = None


# First, try to find the existing Git repository.
# This works both locally and inside a cloned Colab repository.

for path in possible_roots:

    if (
        path.name == REPO_NAME
        and (path / ".git").exists()
    ):
        ROOT = path
        break


# ------------------------------------------------------------
# Detect Google Colab
# ------------------------------------------------------------

IN_COLAB = (
    "google.colab" in sys.modules
    and Path("/content").exists()
)


# ------------------------------------------------------------
# If repository was not found, handle Colab
# ------------------------------------------------------------

if ROOT is None and IN_COLAB:

    ROOT = Path("/content") / REPO_NAME

    if not ROOT.exists():

        print("Google Colab environment detected.")
        print("Cloning GCI World 2026 repository...")

        subprocess.run(
            [
                "git",
                "clone",
                REPO_URL,
                str(ROOT)
            ],
            check=True
        )

        print("Repository cloned successfully.")

    else:

        print(
            "Google Colab environment detected."
        )

        print(
            "Repository already available."
        )


# ------------------------------------------------------------
# Local fallback
# ------------------------------------------------------------

if ROOT is None:

    raise FileNotFoundError(
        "Could not locate the GCI World 2026 repository. "
        "Make sure the notebook is inside the repository."
    )


# ------------------------------------------------------------
# Repository directories
# ------------------------------------------------------------

GLOSSARY_DIR = ROOT / "glossary"
PYTHON_DIR = ROOT / "python"
DATA_SCIENCE_DIR = ROOT / "data_science"
MACHINE_LEARNING_DIR = ROOT / "machine_learning"
AI_DIR = ROOT / "ai"
SQL_DIR = ROOT / "sql"
TIME_SERIES_DIR = ROOT / "time_series"
EXAMPLES_DIR = ROOT / "examples"
CHEATSHEETS_DIR = ROOT / "cheatsheets"
DATASETS_DIR = ROOT / "datasets"
RESOURCES_DIR = ROOT / "resources"


# ------------------------------------------------------------
# Ensure directories exist
# ------------------------------------------------------------

for directory in [
    GLOSSARY_DIR,
    PYTHON_DIR,
    DATA_SCIENCE_DIR,
    MACHINE_LEARNING_DIR,
    AI_DIR,
    SQL_DIR,
    TIME_SERIES_DIR,
    EXAMPLES_DIR,
    CHEATSHEETS_DIR,
    DATASETS_DIR,
    RESOURCES_DIR
]:

    directory.mkdir(
        parents=True,
        exist_ok=True
    )


# ------------------------------------------------------------
# Environment information
# ------------------------------------------------------------

print("==============================================")
print("GCI World 2026 — Learning Library")
print("==============================================")

if IN_COLAB:
    environment = "Google Colab"
else:
    environment = "VS Code / Local"

print(f"Environment : {environment}")
print(f"Project root: {ROOT.resolve()}")
print("Configuration loaded successfully.")
````

````python
# ============================================================
# GCI WORLD 2026 — MULTILINGUAL GLOSSARY
# ============================================================

glossary_file = GLOSSARY_DIR / "glossary.csv"

if glossary_file.exists():
    print(f"Glossary source found: {glossary_file}")
else:
    print("⚠️ Glossary source file not found.")
    print(f"Expected location: {glossary_file}")
````

````python
# ============================================================
# GCI WORLD 2026 — LOAD GLOSSARY
# ============================================================

glossary_df = pd.read_csv(
    glossary_file,
    encoding="utf-8-sig"
)

print("Glossary loaded successfully.")
print(f"Terms: {len(glossary_df)}")

display(glossary_df)
````

````python
# Glossary Search

search_box = widgets.Text(
    description="Search:",
    placeholder="English / Español / 日本語",
    layout=widgets.Layout(width="85%")
)

search_output = widgets.Output()

def search_glossary(change=None):
    query = search_box.value.strip().lower()

    with search_output:
        clear_output(wait=True)

        if not query:
            display(glossary_df)
            return

        mask = glossary_df.astype(str).apply(
            lambda column: column.str.lower().str.contains(query, regex=False)
        ).any(axis=1)

        results = glossary_df[mask]

        if results.empty:
            display(Markdown("⚠️ **No matching terms found.**"))
        else:
            display(results)

search_box.observe(search_glossary, names="value")

display(search_box)
display(search_output)
search_glossary()
````

````python
# ============================================================
# GCI WORLD 2026 — INTERACTIVE LEARNING LIBRARY
# ============================================================

from urllib.parse import quote

GITHUB_BASE_URL = (
    "https://github.com/"
    "GabrielRamos1171998/"
    "GCI-World-2026-September"
)

COLAB_BASE_URL = (
    "https://colab.research.google.com/github/"
    "GabrielRamos1171998/"
    "GCI-World-2026-September/blob/main"
)

class LearningSection:
    """
    Represents a learning area inside the GCI World 2026
    Learning Library.
    """

    def __init__(self, name, description, path, icon="📚"):
        self.name = name
        self.description = description
        self.path = Path(path)
        self.icon = icon

    @property
    def exists(self):
        """Check whether the section exists in the repository."""
        return self.path.exists()

    def get_notebooks(self):
        """Return all notebooks contained in this section."""
        if not self.exists:
            return []
        return sorted(self.path.rglob("*.ipynb"))

    def get_files(self):
        """Return files contained in this section."""
        if not self.exists:
            return []
        return sorted(
            [file for file in self.path.rglob("*") if file.is_file()]
        )

class LearningLibrary:
    """Main navigation manager for the GCI World 2026 Learning Library."""

    def __init__(self):
        self.sections = []

    def add_section(self, section):
        self.sections.append(section)

    def get_section_names(self):
        return [section.name for section in self.sections]

    def get_section(self, name):
        for section in self.sections:
            if section.name == name:
                return section
        return None

    def count_sections(self):
        return len(self.sections)

library = LearningLibrary()

library.add_section(LearningSection(
    "📖 Glossary",
    "Multilingual technical terminology in English, Spanish and Japanese.",
    GLOSSARY_DIR, "📖"
))
library.add_section(LearningSection(
    "🐍 Python",
    "Programming fundamentals, syntax, functions and practical examples.",
    PYTHON_DIR, "🐍"
))
library.add_section(LearningSection(
    "📊 Data Science",
    "Data analysis, visualization, NumPy, Pandas and data cleaning.",
    DATA_SCIENCE_DIR, "📊"
))
library.add_section(LearningSection(
    "🤖 Machine Learning",
    "Machine learning concepts, algorithms, experiments and evaluation.",
    MACHINE_LEARNING_DIR, "🤖"
))
library.add_section(LearningSection(
    "🧠 Artificial Intelligence",
    "AI fundamentals, NLP, LLMs and Retrieval-Augmented Generation.",
    AI_DIR, "🧠"
))
library.add_section(LearningSection(
    "🗄️ SQL",
    "Relational databases, SQL queries, JOINs and data manipulation.",
    SQL_DIR, "🗄️"
))
library.add_section(LearningSection(
    "📈 Time Series",
    "Time-series analysis, patterns and forecasting.",
    TIME_SERIES_DIR, "📈"
))
library.add_section(LearningSection(
    "🧪 Examples",
    "Practical programming, data analysis and machine learning examples.",
    EXAMPLES_DIR, "🧪"
))
library.add_section(LearningSection(
    "📝 Cheat Sheets",
    "Quick references, summaries and study guides.",
    CHEATSHEETS_DIR, "📝"
))
library.add_section(LearningSection(
    "📦 Datasets",
    "Datasets used for analysis, experiments and examples.",
    DATASETS_DIR, "📦"
))
library.add_section(LearningSection(
    "🔗 Resources",
    "External documentation, references and learning materials.",
    RESOURCES_DIR, "🔗"
))

section_menu = widgets.Dropdown(
    options=library.get_section_names(),
    description="Section:",
    layout=widgets.Layout(width="85%")
)

explore_button = widgets.Button(
    description="🚀 Explore Section",
    button_style="primary",
    icon="folder-open"
)

navigation_output = widgets.Output()

def github_file_url(file_path):
    """Create a GitHub URL for a repository file."""
    relative_path = file_path.relative_to(ROOT)
    encoded_path = quote(str(relative_path).replace("\\", "/"), safe="/")
    return f"{GITHUB_BASE_URL}/blob/main/{encoded_path}"

def colab_notebook_url(file_path):
    """Create a Google Colab URL for a notebook."""
    relative_path = file_path.relative_to(ROOT)
    encoded_path = quote(str(relative_path).replace("\\", "/"), safe="/")
    return f"{COLAB_BASE_URL}/{encoded_path}"

def show_section(section):
    """Display information and available resources for the selected section."""
    with navigation_output:
        clear_output(wait=True)

        display(Markdown(f"""
# {section.name}

{section.description}
"""))

        if not section.exists:
            display(Markdown("""
⚠️ **This section has not been created yet.**
"""))
            return

        notebooks = section.get_notebooks()
        files = section.get_files()

        display(Markdown(f"""
### 📂 Section Information

**Repository location:** `{section.path.relative_to(ROOT)}/`

**Notebooks:** {len(notebooks)}

**Files:** {len(files)}
"""))

        if notebooks:
            display(Markdown("### 📓 Available Notebooks"))
            notebook_links = []

            for notebook in notebooks:
                relative_path = notebook.relative_to(ROOT)
                github_url = github_file_url(notebook)
                colab_url = colab_notebook_url(notebook)

                notebook_links.append(f"""
**{notebook.stem}**

`{relative_path}`

[📄 GitHub]({github_url}) · [🔵 Open in Colab]({colab_url})
""")

            display(Markdown("\n\n".join(notebook_links)))
        else:
            display(Markdown("""
### 📓 Available Notebooks

No notebooks have been added to this section yet.

> 🚧 **Content in development**
"""))

        other_files = [
            file for file in files if file.suffix.lower() != ".ipynb"
        ]

        if other_files:
            display(Markdown("### 📦 Other Resources"))
            resource_links = []

            for file in other_files:
                relative_path = file.relative_to(ROOT)
                github_url = github_file_url(file)
                resource_links.append(f"- [{relative_path.name}]({github_url})")

            display(Markdown("\n".join(resource_links)))

def update_section(change):
    """Update the preview when the selected learning section changes."""
    if change["name"] != "value":
        return
    section = library.get_section(change["new"])
    if section:
        show_section(section)

def explore_section(button):
    """Explore the currently selected section."""
    section = library.get_section(section_menu.value)
    if section:
        show_section(section)

section_menu.observe(update_section, names="value")
explore_button.on_click(explore_section)

display(Markdown("""
# 🧭 Explore the Learning Library

Select a learning area to explore its notebooks, datasets and resources.

---

### 📚 Main Learning Areas
"""))

display(section_menu)
display(explore_button)
display(navigation_output)

show_section(library.get_section(section_menu.value))
````
