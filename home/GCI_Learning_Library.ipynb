```python
# Load the glossary from the repository
glossary = pd.read_csv(
    GLOSSARY_DIR / "glossary.csv"
)

print("Glossary loaded successfully.")
print(f"Number of terms: {len(glossary)}")

display(
    glossary[
        [
            "id",
            "category",
            "term_en",
            "term_es",
            "term_ja",
            "reading",
            "level"
        ]
    ]
)
```

```python
# ============================================================
# GLOSSARY SEARCH
# ============================================================

search_box = widgets.Text(
    placeholder="Search in English, Español or 日本語...",
    description="🔎 Search:",
    layout=widgets.Layout(width="80%")
)

category_dropdown = widgets.Dropdown(
    options=["All"] + sorted(glossary["category"].unique().tolist()),
    description="Category:",
    layout=widgets.Layout(width="60%")
)

output = widgets.Output()


def search_glossary(change=None):
    query = search_box.value.strip().lower()
    category = category_dropdown.value

    with output:
        clear_output()

        results = glossary.copy()

        if category != "All":
            results = results[
                results["category"] == category
            ]

        if query:
            mask = (
                results["term_en"].str.lower().str.contains(query, na=False)
                | results["term_es"].str.lower().str.contains(query, na=False)
                | results["term_ja"].str.lower().str.contains(query, na=False)
                | results["reading"].str.lower().str.contains(query, na=False)
                | results["definition_en"].str.lower().str.contains(query, na=False)
                | results["definition_es"].str.lower().str.contains(query, na=False)
            )

            results = results[mask]

        if results.empty:
            display(Markdown("### ❌ No terms found."))
            return

        for _, term in results.iterrows():

            display(
                Markdown(
                    f"""
## {term['term_en']}

**🇬🇧 English:** {term['term_en']}  
**🇪🇸 Español:** {term['term_es']}  
**🇯🇵 日本語:** {term['term_ja']}（{term['reading']}）

**Category:** `{term['category']}`  
**Level:** `{term['level']}`

### Definition

**English:**  
{term['definition_en']}

**Español:**  
{term['definition_es']}

**日本語:**  
{term['definition_ja']}

---
"""
                )
            )


search_box.observe(search_glossary, names="value")
category_dropdown.observe(search_glossary, names="value")

display(search_box)
display(category_dropdown)
display(output)

search_glossary()
```
