```python
dataset_path = Path(path)

csv_files = list(dataset_path.glob("*.csv"))

for file in csv_files:
    print(file.name)
```

```python
data1["Initial Cost Numeric"] = (
    data1["Initial Cost"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

data1["Initial Cost Numeric"] = pd.to_numeric(
    data1["Initial Cost Numeric"],
    errors="coerce"
)
```

```python
data1[
    ["Initial Cost", "Initial Cost Numeric"]
].head(10)
```

```python
data1["Initial Cost Numeric"].describe()
```

```python
mean_cost = data1["Initial Cost Numeric"].mean()
median_cost = data1["Initial Cost Numeric"].median()
std_cost = data1["Initial Cost Numeric"].std()

print("Mean:", mean_cost)
print("Median:", median_cost)
print("Standard deviation:", std_cost)
```

```python
data1["Initial Cost Numeric"].hist(
    figsize=(12, 6),
    bins=30,
    edgecolor="black"
)

plt.title("Distribution of Initial Construction Cost")
plt.xlabel("Initial Cost")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=data1,
    x="Borough",
    y="Initial Cost Numeric"
)

plt.title("Initial Cost by Borough")
plt.xlabel("Borough")
plt.ylabel("Initial Cost")

plt.xticks(rotation=45)
plt.show()
```

```python
borough_cost = (
    data1
    .groupby("Borough")["Initial Cost Numeric"]
    .mean()
    .sort_values(ascending=False)
)

borough_cost
```

```python
borough_cost.plot(
    kind="bar",
    figsize=(12, 6),
    edgecolor="black"
)

plt.title("Average Initial Cost by Borough")
plt.xlabel("Borough")
plt.ylabel("Average Initial Cost")

plt.xticks(rotation=45)
plt.show()
```

```python
job_type_cost = (
    data1
    .groupby("Job Type")["Initial Cost Numeric"]
    .mean()
    .sort_values(ascending=False)
)

job_type_cost
```

```python
job_type_cost.plot(
    kind="bar",
    figsize=(10, 6),
    edgecolor="black"
)

plt.title("Average Initial Cost by Job Type")
plt.xlabel("Job Type")
plt.ylabel("Average Initial Cost")

plt.xticks(rotation=45)
plt.show()
```
